```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

import joblib
```

```python
# Load ML-ready dataset
ml_data = pd.read_csv("data/processed/ml_ready_data.csv")

print("Shape:", ml_data.shape)
print(ml_data.columns)
ml_data.head()
```

```python
# ===============================
# 2. CREATE GAP FEATURE (RUN HERE)
# ===============================

ml_data['gap_5_17'] = (
    ml_data['age_5_17'] /
    (ml_data['age_18_greater'] + 1)
)

display(
    ml_data[['age_5_17', 'age_18_greater', 'gap_5_17']].head()
)
```

```python
# ===============================
# 3. CREATE PROPER RISK SCORE
# ===============================

ml_data['risk_score_5_17'] = (
    ml_data['age_5_17'] /
    (
        ml_data['age_0_5'] +
        ml_data['age_5_17'] +
        ml_data['age_18_greater'] +
        1
    )
)

display(
    ml_data[['age_0_5', 'age_5_17', 'age_18_greater', 'risk_score_5_17']].head(10)
)
```

```python
# ===============================
# 4. ASSIGN RISK LEVELS
# ===============================

def risk_label(score):
    if score >= 0.5:
        return "High Risk"
    elif score >= 0.25:
        return "Medium Risk"
    else:
        return "Low Risk"

ml_data['risk_level_5_17'] = ml_data['risk_score_5_17'].apply(risk_label)

display(
    ml_data[['risk_score_5_17', 'risk_level_5_17']].head(10)
)
```

```python
# ==============================
# FINAL CELL: Save Risk Scores
# ==============================

ml_data.to_csv(
    "data/processed/final_risk_scores.csv",
    index=False
)

print("Final risk dataset saved successfully")
```
